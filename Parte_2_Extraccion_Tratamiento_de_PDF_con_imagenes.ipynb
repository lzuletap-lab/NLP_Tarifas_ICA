# ================ Código para varios PDF =======================

```python
import easyocr
import numpy as np
import gc
import re
import json
import pandas as pd # Herramienta para crear tablas (DataFrames) y exportar a Excel
from pathlib import Path
from pdf2image import convert_from_path, pdfinfo_from_path

def limpiar_texto(texto):
    if not texto:
        return ""
    texto = texto.replace("\n", " ").replace("\r", " ")
    texto = re.sub(r"\s+", " ", texto)
    texto = texto.replace("(", "").replace(")", "").replace("\u2014", " ")
    return texto.strip()

# NUEVO: Función para buscar y extraer la estructura CIIU | Actividad | Tarifa
def extraer_datos_ica(texto_seccion, nombre_municipio):
    datos = []
    # Busca patrones: [4 dígitos] [Texto descriptivo] [Número de tarifa]
    patron_tabla = re.compile(r'\b(\d{4})\s+(.+?)\s+(\d{1,2}(?:[,.]\d+)?)\b')

    for coincidencia in patron_tabla.finditer(texto_seccion):
        datos.append({
            "Municipio": nombre_municipio,
            "CIIU": coincidencia.group(1).strip(),
            "Actividad": coincidencia.group(2).strip(),
            "Tarifa": coincidencia.group(3).strip()
        })
    return datos

print("✅ Librerías y funciones cargadas correctamente.")
```

```python
# 1. Le indicamos la carpeta principal donde están guardados todos tus PDFs
carpeta_pdfs = Path("/content/drive/MyDrive/Maestría Analítica de Datos/PLN/Estatutos municipales/Antioquia/Pdf de imágenes")  ## Aquí debe cambiar la ruta donde se encuentran los PDF del archivo .zip que descargó antes de ejecutar el código

# 2. Inicializamos la Inteligencia Artificial
print("Iniciando motor de IA con GPU...")
lector_ia = easyocr.Reader(['es'], gpu=True)
print("✅ Motor OCR listo y directorio configurado.")
```

```python
# 3. El programa busca todos los archivos que terminen en ".pdf" dentro de la carpeta
for ruta_pdf in carpeta_pdfs.glob("*.pdf"):

    nombre_archivo = ruta_pdf.stem
    print(f"\n--- Iniciando extracción de: {ruta_pdf.name} ---")

    texto_total = ""
    info_pdf = pdfinfo_from_path(str(ruta_pdf))
    total_paginas = info_pdf["Pages"]

    # 4. Procesa todas las páginas de este PDF específico
    for numero_pagina in range(1, total_paginas + 1):
        print(f"Leyendo página {numero_pagina} de {total_paginas}...")

        pagina_actual = convert_from_path(str(ruta_pdf), dpi=300, first_page=numero_pagina, last_page=numero_pagina)
        imagen_matriz = np.array(pagina_actual[0])

        textos_detectados = lector_ia.readtext(
            imagen_matriz,
            detail=0,
            paragraph=True,
            adjust_contrast=True,
            width_ths=0.7
        )

        texto_unido = "\n".join(textos_detectados)
        texto_total += limpiar_texto(texto_unido) + "\n\n"

        del pagina_actual
        del imagen_matriz
        gc.collect()

    # 5. Búsqueda inteligente de las tarifas del ICA dentro del texto extraído
    print("Buscando y estructurando tarifas del impuesto ICA...")
    patron_ica = re.search(r'(?i)IMPUESTO\s+DE\s+INDUSTRIA\s+Y\s+COMERCIO.*?(TARIFA.*?)(?=PARÁGRAFO|ARTÍCULO|\Z)', texto_total, re.DOTALL)

    tarifas_estructuradas = []
    # Extrae el nombre del municipio (ej: "Une" de "Une_ACUERDO...")
    municipio = nombre_archivo.split("_")[0] if "_" in nombre_archivo else nombre_archivo

    if patron_ica:
        texto_ica = patron_ica.group(1)
        tarifas_estructuradas = extraer_datos_ica(texto_ica, municipio)
        print(f"Se estructuraron {len(tarifas_estructuradas)} tarifas ICA detectadas.")

        # Guarda la tabla en un Excel en la misma carpeta
        if tarifas_estructuradas:
            df_tarifas = pd.DataFrame(tarifas_estructuradas)
            ruta_excel = carpeta_pdfs / f"{nombre_archivo}_tarifas.xlsx"
            df_tarifas.to_excel(ruta_excel, index=False)
    else:
        print("No se logró aislar la tabla de tarifas ICA en este documento.")

    # 6. Preparamos el nombre exacto y la ruta para el nuevo archivo JSON
    ruta_json = carpeta_pdfs / f"{nombre_archivo}.json"

    # 7. Organizamos la información como un diccionario (incluyendo las tarifas estructuradas)
    datos_guardar = {
        "nombre_original": ruta_pdf.name,
        "municipio": municipio,
        "tarifas_ica": tarifas_estructuradas,
        "texto_extraido": texto_total
    }

    # 8. Escribimos el archivo JSON
    with open(ruta_json, "w", encoding="utf-8") as archivo_json:
        json.dump(datos_guardar, archivo_json, ensure_ascii=False, indent=4)

    print(f"¡Extracción guardada con éxito en: {ruta_json.name}!")

print("\nPROCESO TOTAL TERMINADO. Se extrajo y estructuró la información de todos los PDFs.")
```

El resultado se valida en la misma carpeta donde se encuentran los PDF originals.
